```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img1 = cv2.imread('XRay_1.jpg')
img2 = cv2.imread('Xray_2.jpg')
```

```python
print(img1.shape)
print(img2.shape)
```

```python
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2_rgb = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img1_rgb)
plt.title("Image 1")

plt.subplot(1,2,2)
plt.imshow(img2_rgb)
plt.title("Image 2")

plt.show()
```

```python
gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(gray1, cmap='gray')
plt.title("Gray Image 1")

plt.subplot(1,2,2)
plt.imshow(gray2, cmap='gray')
plt.title("Gray Image 2")

plt.show()
```

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.hist(gray1.ravel(), bins=256, range=[0,256])
plt.title("Histogram - Image 1")

plt.subplot(1,2,2)
plt.hist(gray2.ravel(), bins=256, range=[0,256])
plt.title("Histogram - Image 2")

plt.show()
```

```python
resized1 = cv2.resize(img1, (300,300))
resized2 = cv2.resize(img2, (300,300))

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(resized1, cv2.COLOR_BGR2RGB))
plt.title("Resized Image 1")

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(resized2, cv2.COLOR_BGR2RGB))
plt.title("Resized Image 2")

plt.show()
```

```python
edges1 = cv2.Canny(gray1, 100, 200)
edges2 = cv2.Canny(gray2, 100, 200)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(edges1, cmap='gray')
plt.title("Edges - Image 1")

plt.subplot(1,2,2)
plt.imshow(edges2, cmap='gray')
plt.title("Edges - Image 2")

plt.show()
```

```python
noise = np.random.normal(0, 25, gray1.shape)

noisy_img = gray1 + noise

noisy_img = np.clip(noisy_img, 0, 255).astype(np.uint8)

plt.imshow(noisy_img, cmap='gray')
plt.title("Noisy Image")
plt.show()
```

```python
filtered_img = cv2.GaussianBlur(noisy_img, (5,5), 0)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(noisy_img, cmap='gray')
plt.title("Noisy Image")

plt.subplot(1,2,2)
plt.imshow(filtered_img, cmap='gray')
plt.title("After Gaussian Filter")

plt.show()
```
